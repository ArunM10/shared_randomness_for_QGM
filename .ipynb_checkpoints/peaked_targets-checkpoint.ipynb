```python
import numpy as np


def fwht_complex(a):
    """
    Fast Walsh-Hadamard transform.
    Input length must be a power of two.
    No normalization is applied.
    """
    a = np.asarray(a, dtype=np.complex128).copy()
    n = a.shape[0]
    h = 1
    while h < n:
        for i in range(0, n, 2 * h):
            x = a[i:i+h].copy()
            y = a[i+h:i+2*h].copy()
            a[i:i+h] = x + y
            a[i+h:i+2*h] = x - y
        h *= 2
    return a


def random_iqp_pdf(num_bits, seed=None, two_body="all_to_all"):
    """
    Generates the exact Born distribution of a random IQP-like circuit:

        |0>^n -- H^n -- exp(i diagonal Z/Z_i Z_j phases) -- H^n -- measure Z.

    This is a small-n statevector/FWHT simulator.
    """
    rng = np.random.default_rng(seed)
    n = int(num_bits)
    N = 1 << n

    x = np.arange(N, dtype=np.uint64)

    # z_i(x) = +1 for bit 0, -1 for bit 1
    z = np.empty((N, n), dtype=np.int8)
    for i in range(n):
        z[:, i] = 1 - 2 * ((x >> i) & 1).astype(np.int8)

    h = rng.uniform(0.0, 2 * np.pi, size=n)

    phase = z @ h

    if two_body == "all_to_all":
        for i in range(n):
            for j in range(i + 1, n):
                Jij = rng.uniform(0.0, 2 * np.pi)
                phase += Jij * z[:, i] * z[:, j]

    elif two_body == "nearest_neighbor_open":
        for i in range(n - 1):
            Jij = rng.uniform(0.0, 2 * np.pi)
            phase += Jij * z[:, i] * z[:, i + 1]

    else:
        raise ValueError("two_body must be 'all_to_all' or 'nearest_neighbor_open'.")

    # State after diagonal IQP phase on uniform superposition
    diag_state = np.exp(1j * phase)

    # Final H^n. Since input amplitude was 1/sqrt(N)
    # and final H contributes another 1/sqrt(N), amplitudes = FWHT(diag_state) / N.
    amplitudes = fwht_complex(diag_state) / N

    pdf = np.abs(amplitudes) ** 2
    pdf = pdf / pdf.sum()
    return pdf


def peaked_hard_candidate_pdf(num_bits, peak_weight=None, peak_index=0,
                              seed=None, two_body="all_to_all"):
    """
    Constructs

        P(x*) = delta,
        P(x)  = (1-delta) Q_C(x)/(1-Q_C(x*)) for x != x*,

    where Q_C is a random IQP-like Born distribution.

    This is peaked and plausibly hard if Q_C is from a hard IQP/RQC family.
    """
    n = int(num_bits)
    N = 1 << n

    if peak_weight is None:
        peak_weight = 1.0 / n

    if not (0.0 < peak_weight < 1.0):
        raise ValueError("peak_weight must lie in (0,1).")

    # if not (0 <= peak_index < N):
    #     raise ValueError("peak_index must be in [0, 2**num_bits).")

    q = random_iqp_pdf(n, seed=seed, two_body=two_body)

    residual = q.copy()
    residual[peak_index] = 0.0
    residual_sum = residual.sum()

    if residual_sum <= 0:
        raise RuntimeError("Residual distribution has zero mass.")

    residual /= residual_sum

    p = (1.0 - 2*peak_weight) * residual
    p[peak_index] = peak_weight

    return p
```

```python
n=6
pg = peaked_hard_candidate_pdf(
    num_bits=n,
    peak_weight=1/10,
    peak_index=[0,63],
    seed=123,
    two_body="all_to_all"
)

samples=10000
def sample_target_function():
    x=range(2**n)
    ret = np.random.choice(x,samples,p=pg)
    return ret.copy()

import matplotlib.pyplot as plt
opt_hist=plt.hist(sample_target_function(),bins=2**n)
```

```python
import numpy as np


def sparse_peaked_pdf(num_bits, support_size=None, seed=None,
                      weight_distribution="exponential",
                      return_support=False):
    """
    Generates a sparse/peaked pseudo-independent distribution.

    Support A has size K = support_size.
    Weights are random positive numbers on A and zero elsewhere.

    For large num_bits this dense vector is impossible to store.
    Use only for modest num_bits.
    """
    rng = np.random.default_rng(seed)
    n = int(num_bits)
    N = 1 << n

    if support_size is None:
        support_size = n**2

    K = int(support_size)

    if K <= 0 or K > N:
        raise ValueError("support_size must satisfy 1 <= K <= 2**num_bits.")

    support = rng.choice(N, size=K, replace=False)

    if weight_distribution == "exponential":
        weights = rng.exponential(scale=1.0, size=K)
    elif weight_distribution == "gamma":
        weights = rng.gamma(shape=1.0, scale=1.0, size=K)
    elif weight_distribution == "uniform":
        weights = rng.random(K)
    else:
        raise ValueError("Unknown weight_distribution.")

    weights = weights / weights.sum()

    p = np.zeros(N, dtype=float)
    p[support] = weights

    if return_support:
        return p, support, weights
    return p
```

```python
n=8
pg = sparse_peaked_pdf(
    num_bits=n
)

samples=10000
def sample_target_function():
    x=range(2**n)
    ret = np.random.choice(x,samples,p=pg)
    return ret.copy()

import matplotlib.pyplot as plt
opt_hist=plt.hist(sample_target_function(),bins=2**n)
```

```python
import numpy as np


def peaked_branch_pdf(q_branch, delta, peak_index):
    """
    Construct a peaked branch distribution

        P_s(x) = delta * 1[x = x_star] 
               + (1 - delta) * Q_s^not_peak(x),

    where Q_s^not_peak is Q_s with the peak_index removed and renormalized.

    Parameters
    ----------
    q_branch : array_like
        Born distribution Q_s(x) for a given branch s.
    delta : float
        Peak weight. Must satisfy 0 < delta < 1.
    peak_index : int
        Integer index of x_star.

    Returns
    -------
    p_branch : np.ndarray
        Peaked branch distribution P_s(x).
    """
    q = np.asarray(q_branch, dtype=float).copy()

    if q.ndim != 1:
        raise ValueError("q_branch must be a one-dimensional probability vector.")

    if np.any(q < -1e-12):
        raise ValueError("q_branch contains negative probabilities.")

    q = np.maximum(q, 0.0)
    q_sum = q.sum()

    if q_sum <= 0:
        raise ValueError("q_branch has zero total mass.")

    q /= q_sum

    if not (0.0 < delta < 1.0):
        raise ValueError("delta must lie in (0, 1).")

    N = len(q)

    if not (0 <= peak_index < N):
        raise ValueError("peak_index is out of range.")

    residual = q.copy()
    residual[peak_index] = 0.0

    residual_sum = residual.sum()

    if residual_sum <= 0:
        raise ValueError(
            "Q_s has all its mass on peak_index, so the residual cannot be normalized."
        )

    residual /= residual_sum

    p = (1.0 - delta) * residual
    p[peak_index] = delta

    return p


def mixed_two_branch_peaked_pdf(q_s0, q_s1, delta_1, delta_2, peak_index_s0, peak_index_s1=None,
                                p1=0.5, p2=0.5):
    """
    Construct

        P'(x) = p1 P_0(x) + p2 P_1(x),

    where

        P_0(x) = delta * 1[x = x*_0] + (1-delta) Q_{s=0,C}^not_peak(x),
        P_1(x) = delta * 1[x = x*_1] + (1-delta) Q_{s=1,C}^not_peak(x).

    Crucially, Q_{s=1,C} is supplied independently. It is NOT assumed to be
    Q_{s=0,C}(bar{x}) or any trivial postprocessing of Q_{s=0,C}.

    Parameters
    ----------
    q_s0 : array_like
        Born distribution Q_{s=0,C}(x).
    q_s1 : array_like
        Born distribution Q_{s=1,C}(x), from the nontrivial angle-flipped branch.
    delta : float
        Peak weight inside each branch.
    peak_index_s0 : int
        Peak location x*_0 for branch s=0.
    peak_index_s1 : int or None
        Peak location x*_1 for branch s=1. If None, uses peak_index_s0.
    p1, p2 : float
        Classical mixing probabilities. They are automatically normalized.

    Returns
    -------
    p_mix : np.ndarray
        Final mixed distribution P'(x).
    p_s0 : np.ndarray
        Branch distribution P_0(x).
    p_s1 : np.ndarray
        Branch distribution P_1(x).
    """
    q_s0 = np.asarray(q_s0, dtype=float)
    q_s1 = np.asarray(q_s1, dtype=float)

    if q_s0.shape != q_s1.shape:
        raise ValueError("q_s0 and q_s1 must have the same shape.")

    if peak_index_s1 is None:
        peak_index_s1 = peak_index_s0

    if p1 < 0 or p2 < 0:
        raise ValueError("p1 and p2 must be nonnegative.")

    norm = p1 + p2

    if norm <= 0:
        raise ValueError("At least one of p1, p2 must be positive.")

    p1 = p1 / norm
    p2 = p2 / norm

    p_s0 = peaked_branch_pdf(q_s0, delta=delta_1, peak_index=peak_index_s0)
    p_s1 = peaked_branch_pdf(q_s1, delta=delta_2, peak_index=peak_index_s1)

    p_mix = p1 * p_s0 + p2 * p_s1
    p_mix /= p_mix.sum()

    return p_mix, p_s0, p_s1
```

```python
n = 6
N = 2**n

rng = np.random.default_rng(123)

# Example placeholders for two genuinely different branch Born distributions.
# In your actual case, q_s0 and q_s1 should come from two different circuit branches:
#   s=0: original angles
#   s=1: intermediate correlated angle flips
q_s0 = rng.random(N)
q_s0 /= q_s0.sum()

q_s1 = rng.random(N)
q_s1 /= q_s1.sum()

delta_1 = 0.6 / n
delta_2 = 0.4 / n
peak_index_s0 = 0
peak_index_s1 = 2**n - 1

p_mix, p0, p1 = mixed_two_branch_peaked_pdf(
    q_s0=q_s0,
    q_s1=q_s1,
    delta_1=delta_1,
    delta_2=delta_2,
    peak_index_s0=peak_index_s0,
    peak_index_s1=peak_index_s1,
    p1=0.3,
    p2=0.7,
)

samples=10000
def sample_target_function():
    x=range(2**n)
    ret = np.random.choice(x,samples,p=p_mix)
    return ret.copy()

import matplotlib.pyplot as plt
opt_hist=plt.hist(sample_target_function(),bins=2**n)
```

```python
import numpy as np


def unequal_width_complement_peaks(num_bits, eps0, eps1, weight0=0.5, dtype=np.float64):
    """
    Exact distribution:
        P(x) = weight0 * G_eps0^{0^N}(x)
             + (1 - weight0) * G_eps1^{1^N}(x)

    where
        G_eps^a(x) = prod_i [(1-eps) if x_i=a_i else eps].

    Args:
        num_bits: number of bits N.
        eps0: noise width around 0^N. Must satisfy 0 <= eps0 <= 1.
        eps1: noise width around 1^N. Must satisfy 0 <= eps1 <= 1.
        weight0: mixture weight for peak around 0^N.
        dtype: numpy dtype.

    Returns:
        p: numpy array of length 2**num_bits.
           p[x] is the probability of bitstring x in integer order.
    """

    if num_bits < 1:
        raise ValueError("num_bits must be positive.")

    if not (0.0 <= eps0 <= 1.0):
        raise ValueError("eps0 must be in [0,1].")

    if not (0.0 <= eps1 <= 1.0):
        raise ValueError("eps1 must be in [0,1].")

    if not (0.0 <= weight0 <= 1.0):
        raise ValueError("weight0 must be in [0,1].")

    dim = 2 ** num_bits
    x = np.arange(dim, dtype=np.uint64)

    # Hamming weight of each bitstring x
    hamming_weight = np.array([int(i).bit_count() for i in x], dtype=np.int64)

    w = hamming_weight
    N = num_bits

    # Distance from 0^N is w
    G0 = (1.0 - eps0) ** (N - w) * (eps0 ** w)

    # Distance from 1^N is N - w
    G1 = (1.0 - eps1) ** w * (eps1 ** (N - w))

    p = weight0 * G0 + (1.0 - weight0) * G1
    p = p.astype(dtype)

    # Numerical cleanup
    p /= p.sum()

    return p
```

```python
n = 6
eps0 = 0.05
eps1 = 0.20

p_mix = unequal_width_complement_peaks(
    num_bits=6,
    eps0=0.15,
    eps1=0.25,
    weight0=0.4
)

samples=10000
def sample_target_function():
    x=range(2**n)
    ret = np.random.choice(x,samples,p=p_mix)
    return ret.copy()

import matplotlib.pyplot as plt
opt_hist=plt.hist(sample_target_function(),bins=2**n)
```
